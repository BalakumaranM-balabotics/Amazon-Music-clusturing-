```python

import pandas as pd

def extract_year(x):
    x = str(x).strip()
    if x.isdigit() and len(x) == 4:
        return int(x)
    return pd.to_datetime(x, format="mixed", errors="coerce").year  
amc_df["release_date"] = amc_df["release_date"].apply(extract_year)
```

```python
print(amc_df["release_date"].tail())
```

```python
import matplotlib.pyplot as plt
features = [
    "danceability",
    "energy",
    "loudness",
    "speechiness",
    "acousticness",
    "instrumentalness",
    "liveness",
    "valence",
    "tempo",
    "duration_ms"
]

amc_df[features].hist(
    figsize=(15, 10),
    bins=30
)

plt.tight_layout()
plt.show()
```

```python
#Scaling
from sklearn.preprocessing import StandardScaler

# Select features
features = ["danceability", "energy", "loudness", "speechiness", "acousticness", "instrumentalness", "liveness", "valence", "tempo", "duration_ms"]

X = amc_df[features]

# Create scaler
scaler = StandardScaler()

# Scale the features
X_scaled = scaler.fit_transform(X)
```

```python

# Convert to DataFrame
X_scaled = pd.DataFrame(X_scaled,columns=features,index=amc_df.index)
print(X_scaled.head())
```

```python
print(X_scaled.mean())
print(X_scaled.std())
```

```python
import sklearn

print(sklearn.__version__)
```

```python
from sklearn.cluster import KMeans
wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of clusters(k)')
plt.ylabel('WCSS')
plt.show()
```

```python
from sklearn.metrics import silhouette_score

for k in range(2, 9):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    print(f'K={k}: {score:.3f}') 
```

```python
from sklearn.metrics import davies_bouldin_score

dbi = davies_bouldin_score(X_scaled, kmeans.labels_)

print("Davies-Bouldin Index:", dbi)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import davies_bouldin_score

dbi_scores = []

for k in range(2, 9):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_scaled)

    dbi = davies_bouldin_score(X_scaled, labels)
    dbi_scores.append(dbi)

    print(f"K={k}: DBI={dbi:.4f}")
```

```python
#therefore K is equal to 3 
```

```python
#Creating K-means model and adding a new column called cluster in the csv file
kmeans = KMeans(n_clusters=3, random_state=42)
amc_df['kmeanscluster'] = kmeans.fit_predict(X_scaled)
```

```python
print(amc_df.groupby('kmeanscluster')[features].mean())
```

```python
cluster_means = amc_df.groupby('kmeanscluster')[features].mean()

for cluster in cluster_means.index:
    print(f"\nCluster {cluster}")
    
    for feature in features:
        value = cluster_means.loc[cluster, feature]
        min_value = cluster_means[feature].min()
        max_value = cluster_means[feature].max()

        if value == max_value:
            position = "Highest"
        elif value == min_value:
            position = "Lowest"
        else:
            position = "Middle"

        print(f"{feature}: {position} ({value:.3f})")
```

```python

#Interpretation
#By considering the above results
#--> For Cluster 0 it is Chill and Acoustic 
#--> For Cluster 1 it is Party and Energetic
#--> For Cluster 2 it is Speach heavy, Rap and Spoken-word
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)
plt.scatter(X_pca[:,0], X_pca[:,1], c=amc_df['kmeanscluster'], cmap='viridis')
plt.title('Clusters (PCA)')
plt.show()
```

```python
print(pca.explained_variance_ratio_.sum())
```

```python
#DBSCAN
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=1, min_samples=5)

db = dbscan.fit_predict(X_scaled)

n_clusters = len(set(db)) - (1 if -1 in db else 0)
noise_count = np.sum(db == -1)

print("Number of clusters:", n_clusters)
print("Number of noise points:", noise_count)

amc_df['dbscan_cluster'] = db
```
